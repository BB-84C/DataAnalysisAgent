```python
import numpy as np
import matplotlib.pyplot as plt

# Generate an array of values from -π to π (100 points)
x = np.linspace(-np.pi, np.pi, 100)

# Compute the sine of each x value
y = np.sin(x)

# Create a plot
plt.figure()
plt.plot(x, y)

# Add labels and a title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Sine Wave')

# Display the plot
plt.show()
```
