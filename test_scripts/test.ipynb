```python
from namespace_mngr.namespace_manager import NamespaceManager

# 初始化 NamespaceManager
nm = NamespaceManager()

# 启动一个 runtime
runtime_id = "runtime_test_1"
nm.start_runtime(runtime_id)

# 模拟几次 MATLAB 脚本调用
nm.log_script_call("opensxm_auto", {"path": "F:/data", "fn": "file1.sxm"}, runtime_id)
nm.log_script_call("image_Fourier_Transform", {"pnum": 3}, runtime_id)
nm.log_script_call("image_polybackrow", {"pnum": 2}, runtime_id)

# 结束 runtime 并获取变量变化和日志
diff = nm.end_runtime(runtime_id)

# 打印结果
print("\n=== 新增或修改变量 ===")
for name, info in diff.items():
    print(f"{name} -> class: {info['class']}, size: {info['size']}")

print("\n=== 该 runtime 执行过的 MATLAB 脚本 ===")
runtime_log = nm.runtime_logs.get(runtime_id, [])
for entry in runtime_log:
    print(f"{entry['script']}  参数: {entry['params']}")

print("\n=== 全局执行日志 ===")
for entry in nm.execution_log:
    print(f"{entry['script']}  参数: {entry['params']}")
```
